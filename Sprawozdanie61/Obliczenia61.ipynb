# Wyznaczanie ogniskowych soczewek cienkich za pomocą ławy optycznej

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Wprowadzenie danych

```python
# --- Zadanie 1: Obraz powiększony ---
zadanie1_powiekszony = np.array([
    [190, 13.5, 176.5],
    [170, 13.5, 156.5],
    [150, 14.0, 136.0],
    [130, 14.0, 116.0],
    [110, 14.5, 95.5],
    [90, 15.0, 75.0],
    [70, 16.5, 53.5],
    [50, 20.0, 30.0]
])

# Rozbicie na kolumny
l_z1_pow = zadanie1_powiekszony[:, 0]
a_z1_pow = zadanie1_powiekszony[:, 1]
b_z1_pow = zadanie1_powiekszony[:, 2]


# --- Zadanie 1: Obraz pomniejszony ---
zadanie1_pomniejszony = np.array([
    [190, 176.0, 13.5],
    [170, 155.5, 14.5],
    [150, 135.0, 14.5],
    [130, 115.5, 14.5],
    [110, 94.5, 15.5],
    [90, 74.5, 15.5],
    [70, 52.5, 17.5],
    [50, 29.5, 20.5]
])

# Rozbicie na kolumny
l_z1_pom = zadanie1_pomniejszony[:, 0]
a_z1_pom = zadanie1_pomniejszony[:, 1]
b_z1_pom = zadanie1_pomniejszony[:, 2]


# --- Zadanie 2 ---
zadanie2 = np.array([
    [190, 13.5, 176.0],
    [170, 13.5, 155.5],
    [150, 14.0, 135.0],
    [130, 14.0, 115.5],
    [110, 14.5, 94.5],
    [90, 15.0, 74.5],
    [70, 16.5, 52.5],
    [50, 20.0, 29.5]
])

# Rozbicie na kolumny
l_z2 = zadanie2[:, 0]
a_z2 = zadanie2[:, 1]
b_z2 = zadanie2[:, 2]


# --- Zadanie 3 ---
zadanie3 = np.array([
    [190, 19.0, 168.5],
    [170, 20.0, 148.5],
    [150, 21.0, 127.5],
    [130, 22.5, 108.5],
    [110, 23.5, 87.0],
    [90, 23.5, 65.0],
    [70, 29.0, 42.0]
])

# Rozbicie na kolumny
l_z3 = zadanie3[:, 0]
a_z3 = zadanie3[:, 1]
b_z3 = zadanie3[:, 2]
```

## Obliczenia ogniskowych

```python
f_z1_pow = 1 / (1 / a_z1_pow + 1 / b_z1_pow)
f_z1_pom = 1 / (1 / a_z1_pom + 1 / b_z1_pom)

f_z1_pow_sr = np.mean(f_z1_pow)
f_z1_pom_sr = np.mean(f_z1_pom)

d_z2 = b_z2 - a_z2
f_z2 = (l_z2 - d_z2**2 / l_z2) / 4

f_z2_sr = np.mean(f_z2)

d_z3 = b_z3 - a_z3
f_u = (l_z3 - d_z3**2 / l_z3) / 4
f_z2_trimmed = f_z2[:len(f_u)]
f_z3 = (f_u*f_z2_trimmed)/(f_u-f_z2_trimmed)

f_z3_sr = np.mean(f_z3)

# Wyświetlenie wyników
print("Ogniskowe zadanie 1 (obraz powiększony):", f_z1_pow)
print("Ogniskowa średnia zadanie 1 (obraz powiększony):", f_z1_pow_sr)
print("Ogniskowe zadanie 1 (obraz pomniejszony):", f_z1_pom)
print("Ogniskowa średnia zadanie 1 (obraz pomniejszony):", f_z1_pom_sr)
print("Odległości d zadanie 2:", d_z2)
print("Ogniskowe zadanie 2:", f_z2)
print("Ogniskowa średnia zadanie 2:", f_z2_sr)
print("Ogniskowa zadanie 3 intermediate f_u:", f_u)
print("Odległości d zadanie 3:", d_z3)
print("Ogniskowe zadanie 3:", f_z3)
print("Ogniskowa średnia zadanie 3:", f_z3_sr)
```

## Zadanie 1 - Prosta regresji liniowej

```python
# policz powiększenia
M_z1_pow = b_z1_pow / a_z1_pow
M_z1_pom = b_z1_pom / a_z1_pom

print("Powiększenia M - powiększony:", M_z1_pow)
print("Powiększenia M - pomniejszony:", M_z1_pom)

# dopasowanie prostej M = m*b + c
wsp_M_pow = np.polyfit(b_z1_pow, M_z1_pow, 1)
prosta_M_pow = np.poly1d(wsp_M_pow)

wsp_M_pom = np.polyfit(b_z1_pom, M_z1_pom, 1)
prosta_M_pom = np.poly1d(wsp_M_pom)

print("Współczynniki prostej M(b) - powiększony:", wsp_M_pow)
print("Współczynniki prostej M(b) - pomniejszony:", wsp_M_pom)
```

## Zadanie 1 - Wykresy

```python
plt.figure(figsize=(18, 12))
plt.scatter(b_z1_pow, M_z1_pow, color='blue', label='Dane eksperymentalne (obraz powiększony)')
plt.plot(b_z1_pow, prosta_M_pow(b_z1_pow), color='red', label='Prosta regresji liniowej')
plt.title('Zadanie 1 - Powiększenie soczewki cienkiej (obraz powiększony)')
plt.ylabel('Powiększenie M')
plt.xlabel('Odległość obrazu b (cm)')
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(18, 12))
plt.scatter(b_z1_pom, M_z1_pom, color='blue', label='Dane eksperymentalne (obraz pomniejszony)')
plt.plot(b_z1_pom, prosta_M_pom(b_z1_pom), color='red', label='Prosta regresji liniowej')
plt.title('Zadanie 1 - Powiększenie soczewki cienkiej (obraz pomniejszony)')
plt.ylabel('Powiększenie M')
plt.xlabel('Odległość obrazu b (cm)')
plt.legend()
plt.grid()
plt.show()
```
