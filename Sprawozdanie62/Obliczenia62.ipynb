## Obliczenie średniej wartości prądu dla każdego kąta

```python
I_avg = (I_rosnaco + I_malejaco) / 2

print("Średnie wartości prądu dla każdego kąta:")
for angle, current in zip(alpha, I_avg):
    print(f"Kąt: {angle}°, Średni prąd: {current}")
```

## Obliczenie wartości funkcji (normalizacja dla cos²(α))

```python
I_min = np.min(I_avg)
I_max = np.mean(I_t)
print(f"Minimalna wartość prądu I_min: {I_min}")
print(f"Maksymalna wartość prądu I_max: {I_max}")
I_normalized = (I_avg - I_min) / (I_max - I_min)
cos_squared = np.cos(np.radians(alpha))**2
print("znormalizowane wartości prądu:")
for angle, current in zip(alpha, I_normalized):
    print(f"Kąt: {angle}°, Znormalizowany prąd: {current}")
```

## Niepewności pomiarowe

```python
delta_I_max = np.maximum(np.abs(np.max(I_avg) - I_max), np.abs(I_max - np.min(I_avg)))
delta = delta_I_max / I_max

print(f"max I_avg: {np.max(I_avg)}")
print(f"min I_avg: {np.min(I_avg)}")

print(f"Maksymalna niepewność prądu delta_I_max: {delta_I_max}")

print(f"Niepewność prądu: {delta}")

delta_normalized = delta / (I_max - I_min)
print(f"Niepewność znormalizowanego prądu: {delta_normalized}")
```

## Wykres zależności I(α) oraz porównanie z cos²(α)

```python
plt.figure(figsize=(18, 12))
plt.scatter(alpha, I_normalized, label='Znormalizowany prąd')
plt.plot(alpha, cos_squared, label='cos²(α)', color='red')
plt.title('Prawo Malusa: Zależność znormalizowanego prądu od kąta α')
plt.xlabel('Kąt α (stopnie)')
plt.ylabel('Znormalizowany prąd I')
plt.legend()
plt.grid()
plt.show()
```

## Wykres zależności I(α) oraz porównanie z cos²(α) wraz z paskami błędów

```python
plt.figure(figsize=(18, 12))
plt.errorbar(alpha, I_normalized, yerr=delta_normalized, fmt='o', label='Znormalizowany prąd z niepewnościami', capsize=5)
plt.plot(alpha, cos_squared, label='cos²(α)', color='red')
plt.title('Prawo Malusa: Zależność znormalizowanego prądu od kąta α')
plt.xlabel('Kąt α (stopnie)')
plt.ylabel('Znormalizowany prąd I')
plt.legend()
plt.grid()
plt.show()
```
