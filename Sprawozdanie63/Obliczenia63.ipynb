# Dyfrakcja światła na szczelinie

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Wprowadzenie danych

```python
L = 2 # m

# --- Tabela 1: D6 = 0.017 mm ---
D6_xmax = np.array([23.7, 37.9])
D6_xmin = np.array([31.2, np.nan])
D6 = 0.017  # mm

# --- Tabela 2: D5 = 0.05 mm ---
D5_xmax = np.array([8.5, 13.5, 19.0, 24.7, 29.5, 35.3])
D5_xmin = np.array([10.9, 16.5, 22.1, 27.9, 32.8, np.nan])
D5 = 0.05  # mm

# --- Tabela 3: D4 = 0.1 mm ---
D4_xmax = np.array([4.3, 7.0, 10.2, 14.0, 17.0, 19.5, 22.6, 24.9, 27.4, 30.0])
D4_xmin = np.array([5.8, 8.6, 11.8, 15.4, 18.5, 21.1, 23.8, 26.9, 28.7, np.nan])
D4 = 0.1  # mm

# --- Tabela 4: D3 = 0.2 mm ---
D3_xmax = np.array([1.0, 3.8, 4.7, 6.1, 7.5, 9.1, 10.4, 11.7])
D3_xmin = np.array([2.7, 4.2, 5.4, 6.7, 8.3, 9.5, 10.9, np.nan])
D3 = 0.2  # mm

# --- Tabela 5: D2 = 0.3 mm ---
D2_xmax = np.array([1.5, 2.4, 3.5, 4.4, 5.3, 6.3, 7.2, 8.1])
D2_xmin = np.array([2.0, 3.0, 3.9, 4.9, 5.7, 6.7, 7.7, 8.6])
D2 = 0.3  # mm

# --- Tabela 6: D1 = 0.4 mm ---
D1_xmax = np.array([1.0, 1.7, 2.4, 3.0, 3.7, 4.5, 5.0, 5.7])
D1_xmin = np.array([1.4, 2.0, 2.7, 3.3, 4.1, 4.7, 5.4, 6.1])
D1 = 0.4  # mm
```

## Obliczenia długości fali

```python
lambda_D6_max = np.array([])
lambda_D6_min = np.array([])
for n in range(len(D6_xmax)):
    temp_max = (D6 * 1e-3 * D6_xmax[n] * 1e-2) / ((2 * (n+1) + 1) * L)
    lambda_D6_max = np.append(lambda_D6_max, temp_max)
    if not np.isnan(D6_xmin[n]):
        temp_min = (D6 * 1e-3 * D6_xmin[n] * 1e-2) / ((2 * (n+1) + 1) * L)
        lambda_D6_min = np.append(lambda_D6_min, temp_min)

lambda_D5_max = np.array([])
lambda_D5_min = np.array([])
for n in range(len(D5_xmax)):
    temp_max = (D5 * 1e-3 * D5_xmax[n] * 1e-2) / ((2 * (n+1) + 1) * L)
    lambda_D5_max = np.append(lambda_D5_max, temp_max)
    if not np.isnan(D5_xmin[n]):
        temp_min = (D5 * 1e-3 * D5_xmin[n] * 1e-2) / ((2 * (n+1) + 1) * L)
        lambda_D5_min = np.append(lambda_D5_min, temp_min)

lambda_D4_max = np.array([])
lambda_D4_min = np.array([])
for n in range(len(D4_xmax)):
    temp_max = (D4 * 1e-3 * D4_xmax[n] * 1e-2) / ((2 * (n+1) + 1) * L)
    lambda_D4_max = np.append(lambda_D4_max, temp_max)
    if not np.isnan(D4_xmin[n]):
        temp_min = (D4 * 1e-3 * D4_xmin[n] * 1e-2) / ((2 * (n+1) + 1) * L)
        lambda_D4_min = np.append(lambda_D4_min, temp_min)

lambda_D3_max = np.array([])
lambda_D3_min = np.array([])
for n in range(len(D3_xmax)):
    temp_max = (D3 * 1e-3 * D3_xmax[n] * 1e-2) / ((2 * (n+1) + 1) * L)
    lambda_D3_max = np.append(lambda_D3_max, temp_max)
    if not np.isnan(D3_xmin[n]):
        temp_min = (D3 * 1e-3 * D3_xmin[n] * 1e-2) / ((2 * (n+1) + 1) * L)
        lambda_D3_min = np.append(lambda_D3_min, temp_min)

lambda_D2_max = np.array([])
lambda_D2_min = np.array([])
for n in range(len(D2_xmax)):
    temp_max = (D2 * 1e-3 * D2_xmax[n] * 1e-2) / ((2 * (n+1) + 1) * L)
    lambda_D2_max = np.append(lambda_D2_max, temp_max)
    if not np.isnan(D2_xmin[n]):
        temp_min = (D2 * 1e-3 * D2_xmin[n] * 1e-2) / ((2 * (n+1) + 1) * L)
        lambda_D2_min = np.append(lambda_D2_min, temp_min)

lambda_D1_max = np.array([])
lambda_D1_min = np.array([])
for n in range(len(D1_xmax)):
    temp_max = (D1 * 1e-3 * D1_xmax[n] * 1e-2) / ((2 * (n+1) + 1) * L)
    lambda_D1_max = np.append(lambda_D1_max, temp_max)
    if not np.isnan(D1_xmin[n]):
        temp_min = (D1 * 1e-3 * D1_xmin[n] * 1e-2) / ((2 * (n+1) + 1) * L)
        lambda_D1_min = np.append(lambda_D1_min, temp_min)

print("Długości fali dla D6 (max):", lambda_D6_max)
print("Długości fali dla D6 (min):", lambda_D6_min)
print("Długości fali dla D5 (max):", lambda_D5_max)
print("Długości fali dla D5 (min):", lambda_D5_min)
print("Długości fali dla D4 (max):", lambda_D4_max)
print("Długości fali dla D4 (min):", lambda_D4_min)
print("Długości fali dla D3 (max):", lambda_D3_max)
print("Długości fali dla D3 (min):", lambda_D3_min)
print("Długości fali dla D2 (max):", lambda_D2_max)
print("Długości fali dla D2 (min):", lambda_D2_min)
print("Długości fali dla D1 (max):", lambda_D1_max)
print("Długości fali dla D1 (min):", lambda_D1_min)
```

## Obliczenie średnich długości fali

```python
lambda_D6_avg = np.mean(np.concatenate((lambda_D6_max, lambda_D6_min)))
lambda_D5_avg = np.mean(np.concatenate((lambda_D5_max, lambda_D5_min)))
lambda_D4_avg = np.mean(np.concatenate((lambda_D4_max, lambda_D4_min)))
lambda_D3_avg = np.mean(np.concatenate((lambda_D3_max, lambda_D3_min)))
lambda_D2_avg = np.mean(np.concatenate((lambda_D2_max, lambda_D2_min)))
lambda_D1_avg = np.mean(np.concatenate((lambda_D1_max, lambda_D1_min)))

print("Średnia długość fali dla D6:", lambda_D6_avg)
print("Średnia długość fali dla D5:", lambda_D5_avg)
print("Średnia długość fali dla D4:", lambda_D4_avg)
print("Średnia długość fali dla D3:", lambda_D3_avg)
print("Średnia długość fali dla D2:", lambda_D2_avg)
print("Średnia długość fali dla D1:", lambda_D1_avg)
```

## Obliczenia niepewności pomiarowych

```python
uc_lambda_D6 = np.std(np.concatenate((lambda_D6_max, lambda_D6_min)), ddof=1)
uc_lambda_D5 = np.std(np.concatenate((lambda_D5_max, lambda_D5_min)), ddof=1)
uc_lambda_D4 = np.std(np.concatenate((lambda_D4_max, lambda_D4_min)), ddof=1)
uc_lambda_D3 = np.std(np.concatenate((lambda_D3_max, lambda_D3_min)), ddof=1)
uc_lambda_D2 = np.std(np.concatenate((lambda_D2_max, lambda_D2_min)), ddof=1)
uc_lambda_D1 = np.std(np.concatenate((lambda_D1_max, lambda_D1_min)), ddof=1)

print("Niepewność pomiarowa długości fali dla D6:", uc_lambda_D6)
print("Niepewność pomiarowa długości fali dla D5:", uc_lambda_D5)
print("Niepewność pomiarowa długości fali dla D4:", uc_lambda_D4)
print("Niepewność pomiarowa długości fali dla D3:", uc_lambda_D3)
print("Niepewność pomiarowa długości fali dla D2:", uc_lambda_D2)
print("Niepewność pomiarowa długości fali dla D1:", uc_lambda_D1)
```
